```python
# Dirty value remove

df.replace(["N/A", "Null", "", " "], np.nan, inplace=True)      # Jin cells me [N/A, Null, "", " "] fill ho rakha hoga unhe replace function "NaN" matlab blank se kar dega
```

```python
# df["Review Rating"]=df.groupby("Category")["Review Rating"].transform(lambda x:x.fillna(x.median()))
                                        # OR
df["Review Rating"]=df["Review Rating"].groupby(df["Category"]).transform(lambda x: x.fillna(x.median()))
```

```python
df = df[(df["age"]>=18) & (df["age"]<=100)]
```

```python
# Feature Engineering
```

```python
df.columns
```

```python
# Create Age Group Column

bins = [17,30,50,60,100]
labels = ["Young Adult", "Adult", "Middle Age", "Senior"]

df["age_group"] = pd.cut(df["age"], bins=bins, labels=labels)
```

```python
df.columns
```

```python
df[["age", "age_group"]]
```

```python
# Create purchase_frequency_days column

frequency_maping = {
    "Fortnightly" : 14,
    "Weekly" : 7,
    "Monthly" : 30,
    "Quarterly" : 90,
    "Bi-Weekly" : 14,
    "Annually" : 365,
    "Every 3 Months" : 90
}

df["purchase_frequency_days"] = df["frequency_of_purchases"].map(frequency_maping)
```
